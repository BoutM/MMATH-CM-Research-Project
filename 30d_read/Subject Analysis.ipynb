# Subject Analysis
### This notebook will explore 3 subjects and their associated records

```python
import pandas as pd
from joblib import load
import os
```

```python
directory = '/work/mbouthil/Research_Project/MIMIC-III_data/MIMIC-III_demo'
os.chdir(directory)

df = pd.read("Demo_Master_Dataframe.joblib")
```

```python
# Altering column values to integer (categorical) or float objects
count = 0
for col, col_values in df.items():
    if df[col].dtype == ('O'):
        df[col] = df[col].astype('category').cat.codes
       
    if df[col].dtype == ('<m8[ns]'):
        df[col] = df[col].dt.total_seconds()
        df[col] = df[col]/(60**2*24)
    
    count += 1
    if count == 13:
        break

df['ed_duration'] = df['ed_duration'].clip(lower=0)
df = df[df['age'] <= 100]
df = df[df['age'] >= 1]
df = df[df['los'] <= 30]
df = df[df['admit_duration'] <= 30]
df = df[df['ed_duration'] <= 1]
df['ed_duration'] = df['ed_duration'].clip(lower=0)

display(df.head(10))
```

```python
subject_id = 10104
```

```python
def subject_to_csv(subject_id): 

    '''This function will save all related subjcet_id
    information for the provided subject_id'''

    # Gathering rows
    inputs = df.loc[df['subject_id'] == subject_id, 'inputs'].iloc[0]
    events = df.loc[df['subject_id'] == subject_id, 'events'].iloc[0]
    labs = df.loc[df['subject_id'] == subject_id, 'labs'].iloc[0]
    prescriptions = df.loc[df['subject_id'] == subject_id, 'prescriptions'].iloc[0]
    procedures = df.loc[df['subject_id'] == subject_id, 'procedures'].iloc[0]
    row = df.loc[df['subject_id'] == subject_id].drop(labels=['inputs', 'events', 'labs', 'prescriptions', 'procedures'],
                                                      axis = 1)
    # Saving .csv files
    inputs.to_csv('subject_inputs.csv', index=False)
    events.to_csv('subject_events.csv', index=False)
    labs.to_csv('subject_labs.csv', index=False)
    prescriptions.to_csv('subject_prescriptions.csv', index=False)
    procedures.to_csv('subject_procedures.csv', index=False)
    row.to_csv('Subject_row.csv', index=False)

    display(prescriptions)
```

```python
subject_to_csv(subject_id)
```
